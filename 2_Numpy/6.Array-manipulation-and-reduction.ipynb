```python
import numpy as np
```

```python
# copia directa y mutable
A = np.array([3, 12, 512])
B = A

A[2] = 21
```

```python
print(f"A: {A}")
print(f"B: {B}")
```

```python
# copia limpia, pero consume memoria (a tener en cuenta)
A = np.array([6, 18, 30])
B = A.copy()

A[2] = 36
```

```python
print(f"A: {A}")
print(f"B: {B}")
```

### Tranformar arrays con `.reshape()`

```python
A = np.arange(1, 10)
print("A:\n", A)

A = A.reshape(3,3)
print("A:\n", A)
```

### Aplanar con `flatten()` o `ravel()` 

```python
# Crea un copia independiente
print("A:\n", A.flatten())

# Apunta al mismop espacio de memoria del arrays a transformar
print("A:\n", A.ravel())
```
