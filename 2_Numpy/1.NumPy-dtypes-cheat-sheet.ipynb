# 📜 NumPy Dtypes Cheat Sheet

## 1. 🔍 La Tabla Maestra (Numéricos)

Lo más importante es saber **cuánto espacio ocupan** y **qué rango de números soportan**.

| Tipo de Dato | Código Corto | Bits / Bytes | Rango Aproximado | Uso Ideal |
| --- | --- | --- | --- | --- |
| **Enteros con Signo** |  |  |  |  |
| `np.int8` | `'i1'` | 8 bits (1B) | -128 a 127 | Imágenes en escala de grises, categorías pequeñas. |
| `np.int16` | `'i2'` | 16 bits (2B) | -32,768 a 32,767 | Audio, conteos medios. |
| `np.int32` | `'i4'` | 32 bits (4B) | -2.1 mil millones a 2.1 mil millones | **Estándar** para contadores o IDs. |
| `np.int64` | `'i8'` | 64 bits (8B) | Astronómico () | Timestamps, IDs financieros grandes. |
| **Enteros Sin Signo** |  |  | *(Solo positivos)* |  |
| `np.uint8` | `'u1'` | 8 bits (1B) | 0 a 255 | **Imágenes RGB**, máscaras de bits. |
| `np.uint16` | `'u2'` | 16 bits (2B) | 0 a 65,535 | Índices de array medianos. |
| `np.uint32` | `'u4'` | 32 bits (4B) | 0 a 4.29 mil millones | Índices grandes. |
| `np.uint64` | `'u8'` | 64 bits (8B) | 0 a  | Cifras positivas gigantescas. |
| **Flotantes (Decimales)** |  |  |  |  |
| `np.float16` | `'f2'` | 16 bits (2B) | Baja precisión (3-4 decimales) | Modelos de IA (Deep Learning) para ahorrar VRAM. |
| `np.float32` | `'f4'` | 32 bits (4B) | Precisión media (~7 decimales) | **Estándar** en Ingeniería de Datos y ML. |
| `np.float64` | `'f8'` | 64 bits (8B) | Alta precisión (~15 decimales) | **Default de Python** y cálculos científicos críticos. |
  
  
> **Nota:** `int` en Python puro equivale automáticamente a `np.int64` (o `int32` en sistemas viejos), y `float` equivale a `np.float64`.

## 2. 🛠️ Definición e Inspección

Cómo asignar tipos al crear arrays y cómo verificar qué tienes.



````python
import numpy as np

# Creación explícita (Recomendado)
arr = np.array([1, 2, 3], dtype='int16') 
arr = np.zeros(5, dtype=np.float32)

# Uso de códigos cortos (Shorthands)
arr = np.ones((2,2), dtype='f4')  # 'f4' = float32

# Inspección
print(arr.dtype)         # float32
print(arr.itemsize)      # 4 (bytes por elemento)
print(arr.nbytes)        # Total bytes (itemsize * size)
````

## 3. 🔄 Casting (Conversión de Tipos)

El método rey es `.astype()`. Recuerda que esto **crea una copia** del array en memoria (a menos que uses `copy=False`, lo cual es riesgoso).

````python
arr = np.array([1.9, 2.5, 3.7])

# De Float a Int (TRUNCAMIENTO, no redondeo)
enteros = arr.astype(np.int32)
# Resultado: [1, 2, 3]  <-- ¡Ojo! Perdiste los decimales

# De String a Float
precios = np.array(["10.5", "20.1", "NaN"])
precios_clean = precios.astype(np.float32)

# Casting Seguro (Safe Casting)
# Evita que conviertas un float64 gigante en un int8 minúsculo por error
# np.can_cast(from_dtype, to_dtype)
es_seguro = np.can_cast(np.float64, np.float32) # True (generalmente)
es_seguro = np.can_cast(np.int64, np.int8)      # False (Overflow seguro)
````

## 4. 🔡 Strings y Objetos (Los "Raros")

NumPy no está optimizado para texto, pero lo maneja.

| Tipo | Descripción | Ejemplo | Comportamiento |
| --- | --- | --- | --- |
| **String (`<U3`)** | Unicode de longitud fija. | `dtype='<U10'` | Si intentas guardar un texto de 11 letras en un `U10`, **lo cortará silenciosamente**. |
| **Object (`O`)** | Punteros a objetos Python. | `dtype='O'` | Permite mezclar tipos (`[1, "hola", True]`) pero **pierdes la velocidad** de NumPy. |

## 5. 💡 Tips de Optimización

1. **Reduce la memoria al 50%:** Si tus datos son edades de personas (0-100), no uses el default `int64`. Puedes usar `int8` o `int16`.
2. **Cuidado con el Overflow:**

    ```Python
    a = np.array([127], dtype='int8')
    a + 1 
    # Resultado: -128 (¡Dio la vuelta! Esto es un bug silencioso muy común)
    ```

  
3. **Valores Nulos (`NaN`):** `NaN` (Not a Number) es técnicamente un **flotante**.
*Si tienes un array de enteros `[1, 2, 3]` y quieres poner un nulo, **debes convertir todo el array a `float`** primero. Los arrays de enteros no soportan `NaN`*.
