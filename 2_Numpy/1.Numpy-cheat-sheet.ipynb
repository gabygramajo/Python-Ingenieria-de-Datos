## 1. Fundamentos y Creación de Arrays

El objeto principal es el `ndarray` (n-dimensional array). A diferencia de las listas, todos sus elementos deben ser del mismo tipo.

| Función | Resultado |
| --- | --- |
| `np.array([1, 2, 3])` | Crea un array de 1D (vector). |
| `np.array([[1, 2], [3, 4]])` | Crea un array de 2D (matriz). |
| `np.zeros((3, 4))` | Matriz de  llena de ceros. |
| `np.ones((2, 2))` | Matriz de  llena de unos. |
| `np.arange(0, 10, 2)` | Array de 0 a 10 con saltos de 2 (0, 2, 4, 6, 8). |
| `np.linspace(0, 1, 5)` | 5 números equiespaciados entre 0 y 1. |

---

## 2. Inspección del Array

Antes de procesar, un Data Engineer siempre verifica la "forma" de sus datos.

* **`arr.shape`**: Devuelve una tupla con las dimensiones (ej: `(5, 2)` son 5 filas y 2 columnas).
* **`arr.ndim`**: Número de dimensiones (1 para vectores, 2 para matrices).
* **`arr.dtype`**: Tipo de dato de los elementos (ej: `int64`, `float32`).
* **`arr.size`**: Cantidad total de elementos.

---

## 3. Operaciones Matemáticas (Vectorización)

Esta es la razón por la que usamos NumPy. Olvida los bucles `for`.

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Operaciones elemento a elemento
suma = a + b          # [5, 7, 9]
producto = a * 2      # [2, 4, 6]
potencia = a ** 2     # [1, 4, 9]

# Funciones Universales (ufuncs)
raiz = np.sqrt(a)
logaritmo = np.log(a)

```

### El Concepto de Broadcasting

Si intentas sumar un array de forma  con uno de forma , NumPy "estirará" el más pequeño para que coincidan. Es magia de rendimiento.

## 4. Indexación y Slicing

Es muy similar a las listas de Python, pero extendido a múltiples dimensiones.

* **Acceso:** `arr[fila, columna]`
* **Slicing (Rebanado):** `arr[inicio:fin, columna]`
* **Filtrado Booleano (Poderoso):**
```python
edades = np.array([20, 25, 30, 18, 40])
mayores_30 = edades[edades > 30] # Retorna [40]

```



---

## 5. Estadística y Agregación

Perfecto para los primeros vistazos a la calidad de los datos (Data Profiling).

| Método | Descripción |
| --- | --- |
| `np.mean(arr)` | Media aritmética. |
| `np.median(arr)` | Mediana (ideal para detectar sesgos por outliers). |
| `np.std(arr)` | Desviación estándar. |
| `np.min(arr)` / `np.max(arr)` | Valores mínimos y máximos. |
| `np.sum(arr, axis=0)` | Suma por columnas. |
| `np.sum(arr, axis=1)` | Suma por filas. |

---

## 6. Manipulación de Forma (Reshaping)

Crucial cuando los datos vienen en un formato (ej: lista plana) y los necesitas en otro (ej: matriz para un modelo).

* **`arr.reshape(filas, columnas)`**: Cambia la estructura sin cambiar los datos.
* **`arr.flatten()`**: Convierte cualquier matriz en un vector de una sola dimensión.
* **`np.transpose(arr)`** o **`arr.T`**: Intercambia filas por columnas.
