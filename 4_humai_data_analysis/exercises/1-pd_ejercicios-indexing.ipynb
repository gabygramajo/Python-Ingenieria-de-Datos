d) ¿Cuál era la tasa de desocupación en el período seleccionado para las mujeres? ¿Era más alta o más baja que la de los hombres? (Extra: pensar el resultado usando una función)

Nota: la tasa se calcula como Desocupados / (Ocupados + Desocupados)

```python
"""
Estado - Condición de actividad:
0 = Entrevista individual no realizada
1 = Ocupado
2 = Descoupado
3 = Inactivo
4 = menor de 10 años
"""

mujeres_desocupadas = eph_2[eph_2["sexo"] == 2].query("estado == 2")["pondera"].sum()
mujeres_ocupadas = eph_2[eph_2["sexo"] == 2].query("estado == 1")["pondera"].sum()
tasa_deso_mujeres = mujeres_desocupadas / (mujeres_desocupadas + mujeres_ocupadas)


hombres_desocupados = eph_2[eph_2["sexo"] == 1].query("estado == 2")["pondera"].sum()
hombres_ocupados = eph_2[eph_2["sexo"] == 1].query("estado == 1")["pondera"].sum()
tasa_deso_hombres = hombres_desocupados / (hombres_desocupados + hombres_ocupados)

print(f"Tasa de desocupación en mujeres {tasa_deso_mujeres*100}%")
print(f"Tasa de desocupación en hombres {tasa_deso_hombres*100}%")
```

```python
def get_tasa_desocupacion(sexo:int) -> float:
  desocupacion = eph_2[eph_2["sexo"] == sexo].query("estado == 2")["pondera"].sum()
  ocupacion = eph_2[eph_2["sexo"] == sexo].query("estado == 1")["pondera"].sum()

  return round((desocupacion / (desocupacion + ocupacion))*100, 3)

tasa_mujer = get_tasa_desocupacion(2)
tasa_varon = get_tasa_desocupacion(1)

print(f"Tasa de desocupación en mujeres {tasa_mujer}%")
print(f"Tasa de desocupación en hombres {tasa_varon}%")

if tasa_mujer > tasa_varon:
  print('La tasa de desocupacion es mas alta en: Mujeres')
else:
  print('La tasa de desocupacion es mas alta en: Varones')
```

Otro Ejemplo de resolución

```python
# Estado 1 (Ocupado) y 2 (Desocupado). Ignoramos inactivos (3) y menores (4).
df_pea = eph_2[eph_2['estado'].isin([1, 2])]
df_pea['estado'].value_counts()
```

```python
# Agrupamos por sexo y estado, y sumamos la ponderación
# sin unstruck
agrupado = df_pea.groupby(['sexo', 'estado'])['pondera'].sum()
print(agrupado)
# utilizando unstruck
#remodela un DataFrame moviendo el nivel más interno de un índice
# jerárquico (filas) hacia el eje de las columnas, convirtiendo un formato "largo" o "apilado" en uno "ancho"
agrupado = df_pea.groupby(['sexo', 'estado'])['pondera'].sum().unstack()
agrupado
```

```python
# Renombramos columnas para que sea legible (1=Ocupado, 2=Desocupado)
agrupado.columns = ['ocupados', 'desocupados']
agrupado
```

```python
# Calculamos la tasa vectorizada (para todas las filas a la vez)
agrupado['tasa_desocupacion'] = (agrupado['desocupados'] / (agrupado['ocupados'] + agrupado['desocupados'])) * 100

print(agrupado['tasa_desocupacion'].round(3))

# Lógica para imprimir quién es mayor (automática)
if agrupado.loc[2, 'tasa_desocupacion'] > agrupado.loc[1, 'tasa_desocupacion']:
    print("La tasa es mayor en Mujeres")
else:
    print("La tasa es mayor en Varones")
```
